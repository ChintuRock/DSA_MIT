```python

# RSA Algorithm with User Input

import math

# Input two prime numbers
p = int(input("Enter first prime number (p): "))
q = int(input("Enter second prime number (q): "))

# Calculate n and Euler's totient
n = p * q
phi = (p - 1) * (q - 1)

# Input public exponent
e = int(input("Enter public exponent (e): "))

# Validate e
if not (1 < e < phi) or math.gcd(e, phi) != 1:
    print("Invalid e! It must be coprime to phi and between 1 and phi.")
    raise SystemExit

# Calculate private exponent
d = pow(e, -1, phi)

print("Public Key:", (e, n))
print("Private Key:", (d, n))

# Input message
message = int(input(f"Enter message (0 to {n - 1}): "))

if not 0 <= message < n:
    print("Message must be between 0 and n-1.")
    raise SystemExit

# Encryption
ciphertext = pow(message, e, n)
print("Encrypted Message:", ciphertext)

# Decryption
decrypted = pow(ciphertext, d, n)
print("Decrypted Message:", decrypted)
```

```python

# Diffie-Hellman Key Exchange with User Input

# Public parameters
p = int(input("Enter prime number (p): "))
g = int(input("Enter generator (g): "))

# Private keys of Alice and Bob
a = int(input("Enter Alice's private key: "))
b = int(input("Enter Bob's private key: "))

if p <= 2 or not (1 < g < p) or a <= 0 or b <= 0:
    print("Invalid input!")
    raise SystemExit

# Calculate public keys
A = pow(g, a, p)
B = pow(g, b, p)

print("Alice's Public Key:", A)
print("Bob's Public Key:", B)

# Calculate shared secrets
alice_secret = pow(B, a, p)
bob_secret = pow(A, b, p)

print("Alice's Shared Secret:", alice_secret)
print("Bob's Shared Secret:", bob_secret)

# Verify shared keys
if alice_secret == bob_secret:
    print("Key Exchange Successful!")
else:
    print("Key Exchange Failed!")
```
