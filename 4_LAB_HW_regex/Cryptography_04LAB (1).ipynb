```python
with open("cryptology.txt", "w") as file:
    num = int(input("Enter the number of lines to be appended"))
    for i in range(num):
        statement = input(f"Enter statement {i + 1}: ")
        file.write(statement + "\n")

print("Five statements have been stored in cryptology.txt.")
```

```python
with open("cryptology.txt", "r") as source_file:
    with open("cryptology_no.txt", "w") as dest_file:
        slno = 1
        for i in source_file:
            dest_file.write(str(slno) + " " + i)
            slno+=1
```

```python
with open("cryptology.txt", "r") as file:
    lines = file.readlines()

    num_lines =  len(lines)
    num_words = 0
    num_letters = 0

    for i in lines:
        num_words+=len(i.split())
        num_letters += len(i)

print("Number of lines:",num_lines)
print("Number of words:", num_words)
print("Number of letters:", num_letters)
```

```python
with open("cryptology_no.txt", "r") as file_no:
    lines = file_no.read()

    upc = 0
    lpc = 0
    dig = 0
    spa = 0
    spchr = 0

    for i in (lines):
        if i.isupper():
            upc+=1
        elif i.islower():
            lpc+=1
        elif i.isdigit():
            dig+=1
        elif i == " ":
            spa+=1
        else:
            spchr+=1

print("Upper case letters :", upc)
print("Lower case letters :", lpc)
print("Digits :", dig)
print("Space :", spa)
print("Special characters :", spchr)
    
```
